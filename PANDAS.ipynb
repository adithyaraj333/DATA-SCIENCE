# *loc* and iloc **bold text**
loc-label
iloc-number

```python
data=[23,52,32,35]
s=pd.Series(data,index=["A","B","C","D"])
print(s.loc["B"])
print(s.iloc[3])
```

```python
data=[10,20,30,40,]
s=pd.Series(data,index=["A","B","C","D"])
print(s.loc["A":"B"])
print(s.iloc[:2])
```
